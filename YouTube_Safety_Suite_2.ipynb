```python
# ================================================================
# 🚀 ULTIMATE SUPER REACTION PIPELINE v14.20 (ADVANCED BYPASS & ROBUST DOWNLOAD)
# ================================================================

import sys
import subprocess
import os
import shutil

print("📦 Checking system environment & updating tools...")

def run_cmd(cmd):
    try:
        subprocess.run(cmd, check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    except Exception:
        pass

if shutil.which("aria2c") is None or shutil.which("ffmpeg") is None or shutil.which("node") is None:
    run_cmd(["apt-get", "update", "-qq"])
    run_cmd(["apt-get", "install", "-y", "-qq", "aria2", "nodejs", "ffmpeg", "curl", "fonts-dejavu-core"])

def install_package(package_name, is_url=False):
    try:
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", "--upgrade", package_name], check=True)
    except Exception as e:
        print(f"⚠️ Failed to install {package_name}: {e}")

install_package("https://github.com/yt-dlp/yt-dlp/archive/master.tar.gz", is_url=True)

required_packages = {
    "whisper": "openai-whisper",
    "edge_tts": "edge-tts",
    "google.genai": "google-genai",
    "nest_asyncio": "nest-asyncio",
    "torch": "torch",
    "requests": "requests",
    "tqdm": "tqdm",
    "googleapiclient": "google-api-python-client",
    "google_auth_oauthlib": "google-auth-oauthlib"
}

for module_name, pip_name in required_packages.items():
    try:
        __import__(module_name)
    except ImportError:
        print(f"⚙️ Installing package: {pip_name}...")
        install_package(pip_name)

print("✅ Environment ready!\n")

import glob
import gc
import json
import asyncio
import time
import re
import random
import requests
from pathlib import Path
from tqdm import tqdm

import torch
import whisper
import edge_tts
import nest_asyncio
import warnings
from google.colab import files

from google.auth.transport.requests import Request
from google.oauth2.credentials import Credentials
from google_auth_oauthlib.flow import InstalledAppFlow
from googleapiclient.discovery import build
from googleapiclient.http import MediaFileUpload

try:
    from google import genai
    from google.colab import userdata
except:
    pass

nest_asyncio.apply()
warnings.filterwarnings("ignore")

print("="*70)
print("🚀 ULTIMATE SUPER REACTION PIPELINE v14.20 (ADVANCED BYPASS & ROBUST DOWNLOAD)")
print("="*70)

for f in ["final_output.mp4", "thumbnail.jpg", "transcript.json", "concat_list.txt", "source_audio.wav", "source_video.mp4", "trimmed_source.mp4", "processed_face.mp4", "token.json"]:
    if os.path.exists(f):
        try: os.remove(f)
        except: pass

for seg in glob.glob("seg_*.*") + glob.glob("vo_*.*") + glob.glob("freeze_*.*"):
    try: os.remove(seg)
    except: pass

FACE_VIDEO = "my_face.mp4"
PROCESSED_FACE = "processed_face.mp4"
RAW_VIDEO = "source_video.mp4"
TRIMMED_VIDEO = "trimmed_source.mp4"
RAW_AUDIO = "source_audio.wav"
OUTPUT_VIDEO = "final_output.mp4"
THUMBNAIL = "thumbnail.jpg"
CLIENT_SECRET_FILE = "client_secret.json"

FACE_SIZE = random.randint(180, 230)
FACE_RADIUS = FACE_SIZE // 2
CORNER_CHOICES = ["br", "bl", "tr", "tl"]
SELECTED_CORNER = random.choice(CORNER_CHOICES)

if SELECTED_CORNER == "br":
    OVERLAY_POS = "x=W-w-30:y=H-h-30"
elif SELECTED_CORNER == "bl":
    OVERLAY_POS = "x=30:y=H-h-30"
elif SELECTED_CORNER == "tr":
    OVERLAY_POS = "x=W-w-30:y=30"
else:
    OVERLAY_POS = "x=30:y=30"

GEMINI_API_KEY = "YOUR_GEMINI_API_KEY"
gemini_client = None
GLOBAL_SOURCE_URL = ""
SOURCE_CHANNEL_NAME = "Independent Financial News Broadcast"
ORIGINAL_DESCRIPTION = ""

def initialize_gemini():
    global gemini_client
    try:
        api_key = os.getenv("GEMINI_API_KEY") or GEMINI_API_KEY
        try:
            api_key = userdata.get("GEMINI_API_KEY")
        except:
            pass
        if api_key and api_key != "YOUR_GEMINI_API_KEY":
            gemini_client = genai.Client(api_key=api_key)
            print("🤖 [Gemini AI]: API Connected securely.")
            return True
    except Exception as e:
        print(f"⚠️ Gemini Init Error: {e}")
    print("⚠️ [Gemini AI]: Running in fallback analytical mode.")
    return False

def call_gemini(prompt):
    global gemini_client
    if not gemini_client:
        return None
    for model_name in ["gemini-2.5-flash", "gemini-1.5-flash", "gemini-1.5-pro"]:
        try:
            res = gemini_client.models.generate_content(model=model_name, contents=prompt)
            if res and res.text:
                return res.text.strip()
        except Exception:
            continue
    return None

def get_duration(path):
    if not os.path.exists(path):
        return 0
    cmd = ["ffprobe", "-v", "error", "-show_entries", "format=duration", "-of", "default=noprint_wrappers=1:nokey=1", path]
    try:
        res = subprocess.run(cmd, capture_output=True, text=True, check=True)
        return float(res.stdout.strip())
    except:
        return 0

def fetch_public_proxies():
    proxies = []
    try:
        res = requests.get("https://raw.githubusercontent.com/TheSpeedX/PROXY-List/master/http.txt", timeout=5)
        if res.status_code == 200:
            for line in res.text.splitlines()[:20]:
                line = line.strip()
                if line: proxies.append(f"http://{line}")
    except:
        pass
    return proxies

def get_youtube_metadata_info(video_url):
    global SOURCE_CHANNEL_NAME, ORIGINAL_DESCRIPTION
    try:
        cookie_arg = ["--cookies", "cookies.txt"] if os.path.exists("cookies.txt") else []
        res_channel = subprocess.run(["yt-dlp"] + cookie_arg + ["--print", "%(channel)s", video_url], capture_output=True, text=True, check=True, timeout=10)
        channel = res_channel.stdout.strip()
        if channel and channel != "NA":
            SOURCE_CHANNEL_NAME = channel

        res_desc = subprocess.run(["yt-dlp"] + cookie_arg + ["--print", "%(description)s", video_url], capture_output=True, text=True, check=True, timeout=10)
        desc = res_desc.stdout.strip()
        if desc and desc != "NA":
            ORIGINAL_DESCRIPTION = desc
    except:
        pass

def download_source():
    global GLOBAL_SOURCE_URL, SOURCE_CHANNEL_NAME, ORIGINAL_DESCRIPTION
    print("\n🎬 Video Source Selection:")
    print("1️⃣ Direct YouTube Link (Advanced Bypass & Multi-Client Pool)")
    print("2️⃣ Local File Upload (Manual)")

    choice = input("\nOption select karein (1 ya 2 enter karein): ").strip()

    if choice == "1":
        video_url = input("🔗 Video Link paste karein: ").strip()
        GLOBAL_SOURCE_URL = video_url
        clean_url = video_url.split('&')[0]

        get_youtube_metadata_info(clean_url)
        print(f"📌 Detected Channel Source: {SOURCE_CHANNEL_NAME}")

        success = False
        cookie_arg = []
        if os.path.exists("cookies.txt"):
            print("🍪 Found 'cookies.txt'! Using authenticated bypass...")
            cookie_arg = ["--cookies", "cookies.txt"]

        # Method A: iOS / TV Embed Client Bypass
        print("\n🚀 [1/5 - Method A]: Downloading via iOS/TV Embed Client Bypass...")
        try:
            if os.path.exists(RAW_VIDEO): os.remove(RAW_VIDEO)
            cmd = [
                "yt-dlp", "--extractor-args", "youtube:player_client=ios,tv",
                "-f", "best[ext=mp4]/best", "-o", RAW_VIDEO, "--no-check-certificate"
            ] + cookie_arg + [clean_url]
            subprocess.run(cmd, check=True)
            if os.path.exists(RAW_VIDEO) and get_duration(RAW_VIDEO) > 2:
                success = True
        except:
            pass

        # Method B: Chrome Impersonation + aria2c Speed
        if not success:
            print("⚠️ Method A failed. Switching to Method B (Chrome Impersonation + aria2c)...")
            try:
                if os.path.exists(RAW_VIDEO): os.remove(RAW_VIDEO)
                cmd = [
                    "yt-dlp", "--impersonate", "chrome",
                    "--external-downloader", "aria2c",
                    "--external-downloader-args", "aria2c:'--max-connection-per-server=16 --split=16'",
                    "-f", "best[ext=mp4]/best", "-o", RAW_VIDEO, "--no-check-certificate"
                ] + cookie_arg + [clean_url]
                subprocess.run(cmd, check=True)
                if os.path.exists(RAW_VIDEO) and get_duration(RAW_VIDEO) > 2:
                    success = True
            except:
                pass

        # Method C: Android / Web Client Pool
        if not success:
            print("⚠️ Method B failed. Switching to Method C (Android/Web pool)...")
            try:
                if os.path.exists(RAW_VIDEO): os.remove(RAW_VIDEO)
                cmd = [
                    "yt-dlp", "--extractor-args", "youtube:player_client=android,web",
                    "-f", "best[ext=mp4]/best", "-o", RAW_VIDEO, "--no-check-certificate"
                ] + cookie_arg + [clean_url]
                subprocess.run(cmd, check=True)
                if os.path.exists(RAW_VIDEO) and get_duration(RAW_VIDEO) > 2:
                    success = True
            except:
                pass

        # Method D: Automated Proxy Rotation
        if not success:
            print("⚠️ Method C failed. Switching to Method D (Automated Proxy Rotation)...")
            proxies = fetch_public_proxies()
            random.shuffle(proxies)
            for proxy in proxies[:10]:
                try:
                    if os.path.exists(RAW_VIDEO): os.remove(RAW_VIDEO)
                    cmd = [
                        "yt-dlp", "--proxy", proxy, "--extractor-args", "youtube:player_client=android,ios",
                        "-f", "best[ext=mp4]/best", "-o", RAW_VIDEO, "--no-check-certificate", "--socket-timeout", "10"
                    ] + cookie_arg + [clean_url]
                    subprocess.run(cmd, check=True, timeout=15)
                    if os.path.exists(RAW_VIDEO) and get_duration(RAW_VIDEO) > 2:
                        success = True
                        break
                except:
                    continue

        if not success:
            print("\n❌ Online download block ho gaya hai.")
            print("💡 Tip: Option 2 select kar ke apni video direct upload kar lein, ya video ko apne PC par download kar ke Colab mein drag-and-drop kar dein.")
            raise Exception("❌ Direct download failed. Please use Option 2.")

        print("✅ Direct download completed successfully!")
    else:
        GLOBAL_SOURCE_URL = ""
        SOURCE_CHANNEL_NAME = "Expert Guest Commentary"
        ORIGINAL_DESCRIPTION = "Independent expert financial commentary and market insight breakdown."
        print("\n📁 [1/5]: Apnay PC se video file select karein:")
        uploaded = files.upload()
        if not uploaded:
            raise Exception("❌ Koi file select nahi ki gayi!")
        uploaded_filename = list(uploaded.keys())[0]
        if uploaded_filename != RAW_VIDEO:
            if os.path.exists(RAW_VIDEO): os.remove(RAW_VIDEO)
            os.rename(uploaded_filename, RAW_VIDEO)
        print("✅ Local Video File Successfully Loaded!")

    return True

def process_adaptive_transcript():
    print("\n🤖 [2/5]: Running Whisper Audio Transcription & Precise Transcript Breakdown...")
    device = "cuda" if torch.cuda.is_available() else "cpu"

    cmd_wav = ["ffmpeg", "-y", "-i", RAW_VIDEO, "-ar", "16000", "-ac", "1", "-c:a", "pcm_s16le", RAW_AUDIO]
    subprocess.run(cmd_wav, check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

    model = whisper.load_model("base", device=device)
    result = model.transcribe(RAW_AUDIO, fp16=(device == "cuda"), verbose=False)

    segments = result.get("segments", [])
    total_dur = get_duration(RAW_VIDEO)

    print("✂️ Applying Anti-Branding Crop Shield to remove source channel names...")
    subprocess.run([
        "ffmpeg", "-y", "-i", RAW_VIDEO,
        "-vf", "crop=in_w:in_h-70:0:35,scale=1920:1080",
        "-c:v", "libx264", "-preset", "ultrafast", "-c:a", "copy", TRIMMED_VIDEO
    ], check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

    full_transcript_text = " ".join([seg["text"] for seg in segments])

    num_parts = random.randint(6, 8)
    interval = max(60.0, (total_dur - 120.0) / num_parts)

    break_timestamps = []
    current_t = 30.0
    for _ in range(num_parts):
        if current_t < total_dur - 60:
            break_timestamps.append(float(current_t))
            current_t += interval

    breakdowns = []

    print("🤖 Generating 100% Transformative Financial Commentary for each transcript block...")
    for idx, pause_time in enumerate(tqdm(break_timestamps, desc="🧠 Scripting Segments")):
        segment_text = ""
        for seg in segments:
            if pause_time <= seg["start"] <= pause_time + 45:
                segment_text += seg["text"] + " "

        if not segment_text.strip():
            segment_text = f"Crucial market updates and expert interview details around minute {int(pause_time // 60)}."

        prompt = (
            f"Write an original, deeply analytical financial commentary and expert educational reaction script "
            f"written in a natural, conversational human broadcast tone based on this excerpt: '{segment_text.strip()}'. "
            f"Use professional financial phrasing. Keep it informative, engaging, and detailed enough to sustain a 90 to 120 second discussion."
        )

        script = call_gemini(prompt)
        if not script:
            script = "Examining the broader economic indicators, notice how institutional investors and central banking policies are dynamically altering market trends. Smart capital is continuously realigning to mitigate systemic risks while evaluating long-term portfolio growth."

        segment_end = min(pause_time + 45.0, total_dur)
        breakdowns.append({"part": idx+1, "start": pause_time, "end": segment_end, "script": script})

    return breakdowns, full_transcript_text

async def generate_voiceover_audio(breakdowns):
    print("\n🤖 [3/5]: Synthesizing Extended Voiceover Audio (1.5 to 2 minutes per segment)...")
    chosen_voice = "en-US-AndrewNeural"

    for idx, b in enumerate(tqdm(breakdowns, desc="🗣️ Voiceover TTS")):
        if b.get("script"):
            vo_file = f"vo_{b['part']}.mp3"
            success = False
            extended_script = (b["script"] + " ") * 3

            for attempt in range(3):
                try:
                    rate_val = "+2%"
                    communicate = edge_tts.Communicate(extended_script, chosen_voice, rate=rate_val)
                    await communicate.save(vo_file)

                    audio_dur = get_duration(vo_file)
                    if os.path.exists(vo_file) and audio_dur >= 70:
                        b["audio_file"] = vo_file
                        b["audio_dur"] = audio_dur
                        success = True
                        break
                except Exception:
                    await asyncio.sleep(2)

            if not success:
                try:
                    communicate = edge_tts.Communicate(b["script"] * 2, chosen_voice)
                    await communicate.save(vo_file)
                    b["audio_file"] = vo_file
                    b["audio_dur"] = get_duration(vo_file)
                except:
                    subprocess.run(["ffmpeg", "-y", "-f", "lavfi", "-i", "anullsrc=r=44100:cl=stereo", "-t", "90", vo_file], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
                    b["audio_file"] = vo_file
                    b["audio_dur"] = 90.0

            await asyncio.sleep(0.5)

    print("✅ Extended Voiceovers generated successfully.")

def prepare_anti_reuse_face():
    if not os.path.exists(FACE_VIDEO):
        subprocess.run(["ffmpeg", "-y", "-f", "lavfi", "-i", "color=c=blue:size=220x220", "-t", "30", "-c:v", "libx264", "-pix_fmt", "yuv420p", FACE_VIDEO], check=True)

    hue_val = round(random.uniform(-0.25, 0.25), 2)
    subprocess.run(["ffmpeg", "-y", "-i", FACE_VIDEO, "-vf", f"hue=h={hue_val}*PI", "-c:v", "libx264", "-preset", "ultrafast", "-c:a", "copy", PROCESSED_FACE], check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

def render_segmented_video(breakdowns):
    print(f"\n🤖 [4/5]: Building Pipeline with Freeze-Frame Pacing (Face Overlay at '{SELECTED_CORNER}' corner)...")
    prepare_anti_reuse_face()
    segment_files = []

    use_gpu = False
    try:
        res = subprocess.run(["ffmpeg", "-encoders"], capture_output=True, text=True)
        if "h264_nvenc" in res.stdout:
            use_gpu = True
            print("🚀 GPU Hardware Acceleration (NVENC) Enabled!")
    except:
        pass

    for b in tqdm(breakdowns, desc="🎬 Rendering Precise Segments & Freezes"):
        part, start_t, end_t = b["part"], b["start"], b["end"]
        clip_dur = max(0.1, end_t - start_t)
        seg_video = f"seg_{part}_play.mp4"

        v_codec_args = ["-c:v", "h264_nvenc", "-preset", "p1"] if use_gpu else ["-c:v", "libx264", "-preset", "ultrafast", "-crf", "23"]

        try:
            subprocess.run([
                "ffmpeg", "-y",
                "-ss", str(start_t), "-i", TRIMMED_VIDEO,
                "-i", PROCESSED_FACE,
                "-t", str(clip_dur),
                "-filter_complex", f"[0:v]fps=30,format=yuv420p[v0];[1:v]scale={FACE_SIZE}:{FACE_SIZE},format=yuva420p,geq=lum='p(X,Y)':a='if(lte(hypot(X-{FACE_RADIUS},Y-{FACE_RADIUS}),{FACE_RADIUS}),255,0)'[face];[v0][face]overlay={OVERLAY_POS}[outv]",
                "-map", "[outv]", "-map", "0:a",
                "-ar", "44100", "-ac", "2"
            ] + v_codec_args + [
                "-c:a", "aac", "-b:a", "128k", seg_video
            ], check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
        except:
            fallback_args = ["-c:v", "libx264", "-preset", "ultrafast", "-crf", "23"]
            subprocess.run([
                "ffmpeg", "-y",
                "-ss", str(start_t), "-i", TRIMMED_VIDEO,
                "-i", PROCESSED_FACE,
                "-t", str(clip_dur),
                "-filter_complex", f"[0:v]fps=30,format=yuv420p[v0];[1:v]scale={FACE_SIZE}:{FACE_SIZE},format=yuva420p,geq=lum='p(X,Y)':a='if(lte(hypot(X-{FACE_RADIUS},Y-{FACE_RADIUS}),{FACE_RADIUS}),255,0)'[face];[v0][face]overlay={OVERLAY_POS}[outv]",
                "-map", "[outv]", "-map", "0:a",
                "-ar", "44100", "-ac", "2"
            ] + fallback_args + [
                "-c:a", "aac", "-b:a", "128k", seg_video
            ], check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

        segment_files.append(seg_video)

        if b.get("script") and b.get("audio_file"):
            freeze_video = f"freeze_{part}_breakdown.mp4"
            audio_duration = b.get("audio_dur", 90.0)

            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-ss", str(max(0, end_t - 0.5)), "-i", TRIMMED_VIDEO,
                    "-i", b["audio_file"], "-i", PROCESSED_FACE,
                    "-t", str(audio_duration),
                    "-filter_complex", f"[0:v]loop=loop=-1:size=1:start=0,fps=30,format=yuv420p[v0];[2:v]scale={FACE_SIZE}:{FACE_SIZE},format=yuva420p,geq=lum='p(X,Y)':a='if(lte(hypot(X-{FACE_RADIUS},Y-{FACE_RADIUS}),{FACE_RADIUS}),255,0)'[face];[v0][face]overlay={OVERLAY_POS}[outv]",
                    "-map", "[outv]", "-map", "1:a",
                    "-ar", "44100", "-ac", "2"
                ] + v_codec_args + [
                    "-c:a", "aac", "-b:a", "128k", freeze_video
                ], check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
            except:
                fallback_args = ["-c:v", "libx264", "-preset", "ultrafast", "-crf", "23"]
                subprocess.run([
                    "ffmpeg", "-y",
                    "-ss", str(max(0, end_t - 0.5)), "-i", TRIMMED_VIDEO,
                    "-i", b["audio_file"], "-i", PROCESSED_FACE,
                    "-t", str(audio_duration),
                    "-filter_complex", f"[0:v]loop=loop=-1:size=1:start=0,fps=30,format=yuv420p[v0];[2:v]scale={FACE_SIZE}:{FACE_SIZE},format=yuva420p,geq=lum='p(X,Y)':a='if(lte(hypot(X-{FACE_RADIUS},Y-{FACE_RADIUS}),{FACE_RADIUS}),255,0)'[face];[v0][face]overlay={OVERLAY_POS}[outv]",
                    "-map", "[outv]", "-map", "1:a",
                    "-ar", "44100", "-ac", "2"
                ] + fallback_args + [
                    "-c:a", "aac", "-b:a", "128k", freeze_video
                ], check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

            segment_files.append(freeze_video)

    print("\n🧵 Stitching final video segments together...")
    inputs_cmd = []
    filter_parts = []
    for i, sf in enumerate(segment_files):
        inputs_cmd.extend(["-i", sf])
        filter_parts.append(f"[{i}:v][{i}:a]")

    concat_filter = "".join(filter_parts) + f"concat=n={len(segment_files)}:v=1:a=1[outv][outa]"

    stitch_cmd = [
        "ffmpeg", "-y"
    ] + inputs_cmd + [
        "-filter_complex", concat_filter,
        "-map", "[outv]", "-map", "[outa]"
    ] + v_codec_args + [
        "-c:a", "aac", "-b:a", "128k", OUTPUT_VIDEO
    ]

    try:
        subprocess.run(stitch_cmd, check=True)
    except:
        stitch_cmd = [
            "ffmpeg", "-y"
        ] + inputs_cmd + [
            "-filter_complex", concat_filter,
            "-map", "[outv]", "-map", "[outa]"
        ] + ["-c:v", "libx264", "-preset", "ultrafast", "-crf", "23"] + [
            "-c:a", "aac", "-b:a", "128k", OUTPUT_VIDEO
        ]
        subprocess.run(stitch_cmd, check=True)

    print(f"✅ Video Composition Completed! Final Duration: ~{int(get_duration(OUTPUT_VIDEO)/60)} Mins")

def generate_metadata(transcript_summary):
    print("🤖 Generating Title & Split-Layout Professional Thumbnail (Wise Metals Investor Style)...")

    ai_title = call_gemini(f"Create a professional, engaging YouTube financial podcast title under 80 characters based on this content: '{transcript_summary[:600]}'. Avoid clickbait.") or "Expert Financial Interview & Market Outlook Analysis"
    ai_title = ai_title.strip('"')

    thumb_text = call_gemini(f"Based on this video summary: '{transcript_summary[:400]}', provide a professional YouTube thumbnail headline consisting strictly of 3 to 5 words (e.g., 'GLOBAL MARKET CRASH' or 'GOLD PRICE SURGE'). Return only raw uppercase text.")
    if not thumb_text or len(thumb_text.split()) > 5 or len(thumb_text) > 25:
        thumb_text = "MARKET OUTLOOK SHIFT"

    thumb_text = thumb_text.upper().replace("'", "").replace('"', "").strip()

    total_dur = get_duration(TRIMMED_VIDEO)
    thumb_time = random.uniform(total_dur * 0.35, total_dur * 0.65) if total_dur > 100 else 10.0

    font_path = "/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf"
    if not os.path.exists(font_path):
        font_path = "/usr/share/fonts/truetype/liberation/LiberationSans-Bold.ttf"

    words = thumb_text.split()
    mid = len(words) // 2
    line1 = " ".join(words[:mid]) if len(words) > 2 else words[0]
    line2 = " ".join(words[mid:]) if len(words) > 2 else (words[1] if len(words) > 1 else "")

    filter_complex_thumb = (
        "scale=1280:720,"
        "eq=contrast=1.15:saturation=1.2,"
        "drawbox=x=0:y=0:w=680:h=720:color=black@0.75:t=fill,"
        f"drawtext=fontfile='{font_path}':text='{line1}':fontcolor=white:fontsize=65:borderw=4:bordercolor=black@0.9:x=60:y=240,"
        f"drawtext=fontfile='{font_path}':text='{line2}':fontcolor=yellow:fontsize=65:borderw=4:bordercolor=black@0.9:x=60:y=330"
    )

    subprocess.run([
        "ffmpeg", "-y", "-ss", str(thumb_time), "-i", TRIMMED_VIDEO,
        "-vframes", "1", "-vf", filter_complex_thumb,
        THUMBNAIL
    ], check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

    if not os.path.exists(THUMBNAIL) or os.path.getsize(THUMBNAIL) == 0:
        subprocess.run([
            "ffmpeg", "-y", "-f", "lavfi", "-i", "color=c=0x0f172a:s=1280x720",
            "-vf", f"drawtext=fontfile='{font_path}':text='{thumb_text}':fontcolor=yellow:fontsize=60:borderw=4:bordercolor=black@0.8:x=60:y=300",
            "-vframes", "1", THUMBNAIL
        ], check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

    return ai_title

def upload_to_youtube(video_path, thumbnail_path, title, transcript_text, privacy_status="private"):
    global GLOBAL_SOURCE_URL, SOURCE_CHANNEL_NAME, ORIGINAL_DESCRIPTION
    print("\n📤 [5/5]: Uploading Podcast to YouTube with 100% Unique Transcript & Dynamic Source Credit...")

    if not os.path.exists(CLIENT_SECRET_FILE):
        print(f"⚠️ '{CLIENT_SECRET_FILE}' file nahi mili! YouTube upload skip kiya ja raha hai.")
        return

    SCOPES = ["https://www.googleapis.com/auth/youtube.upload"]
    credentials = None

    try:
        if os.path.exists("token.json"):
            os.remove("token.json")

        flow = InstalledAppFlow.from_client_secrets_file(CLIENT_SECRET_FILE, SCOPES)
        flow.redirect_uri = 'urn:ietf:wg:oauth:2.0:oob'
        auth_url, _ = flow.authorization_url(prompt='consent', access_type='offline', include_granted_scopes='false')

        print("\n" + "="*70)
        print("🔑 GOOGLE YOUTUBE ACCOUNT AUTHORIZATION REQUIRED:")
        print(f"1. Is link ko apnay browser mein kholen:\n{auth_url}")
        print("2. Google account se sign in kar ke access allow karein.")
        print("3. Wahan jo authorization code milega, usay copy kar ke yahan paste karein.")
        print("="*70 + "\n")

        code = input("🔑 Verification Code yahan paste karein: ").strip()
        flow.fetch_token(code=code)
        credentials = flow.credentials

        with open("token.json", "w") as token:
            token.write(credentials.to_json())
        print("✅ OAuth credentials successfully saved!")

        youtube = build("youtube", "v3", credentials=credentials)

        adapted_desc = call_gemini(
            f"Write a completely unique, highly professional YouTube podcast video description by synthesizing the original source description: '''{ORIGINAL_DESCRIPTION[:1200]}''' "
            f"with the unique transcript text extracted from this specific video: '''{transcript_text[:1200]}'''. "
            f"Ensure the description specifically highlights the unique talking points of this video, keeping it clean, engaging, and well-structured without repetitive generic text."
        )
        if not adapted_desc:
            adapted_desc = transcript_text[:1000]

        if GLOBAL_SOURCE_URL:
            source_credit_section = f"📌 CREDIT / ORIGINAL SOURCE:\n• Broadcast Channel: {SOURCE_CHANNEL_NAME}\n• Original Source Link: {GLOBAL_SOURCE_URL}"
        else:
            source_credit_section = f"📌 CREDIT / ORIGINAL SOURCE:\n• Broadcast Channel: {SOURCE_CHANNEL_NAME}"

        contact_section = "✉️ For copyright matters or fair use inquiries, contact: silentstorm9705@gmail.com"

        full_description = (
            f"🎙 **Professional Financial & Market Commentary Analysis**\n\n"
            f"{adapted_desc}\n\n"
            f"{source_credit_section}\n\n"
            f"━━━━━━━━━━━━━━━━━━━━━━━\n"
            f"⚖️ **FAIR USE & COMPLIANCE DISCLAIMER:**\n"
            f"Copyright Disclaimer Under Section 107 of the Copyright Act 1976, allowance is made for 'fair use' for purposes such as criticism, comment, news reporting, teaching, scholarship, and research.\n\n"
            f"{contact_section}\n"
            f"━━━━━━━━━━━━━━━━━━━━━━━\n\n"
            f"#Finance #Investing #Macroeconomics #Markets #Economy"
        )

        body = {
            "snippet": {
                "title": title[:100],
                "description": full_description[:5000],
                "tags": ["Finance", "Macroeconomics", "Investing", "Podcast", "Markets"],
                "categoryId": "25"
            },
            "status": {
                "privacyStatus": privacy_status,
                "selfDeclaredMadeForKids": False
            }
        }

        media = MediaFileUpload(video_path, chunksize=-1, resumable=True)
        request = youtube.videos().insert(part="snippet,status", body=body, media_body=media)

        response = None
        while response is None:
            status, response = request.next_chunk()
            if status:
                print(f"📥 Upload Progress: {int(status.progress() * 100)}%")

        video_id = response.get("id")
        print(f"✅ Video successfully uploaded! Video ID: {video_id}")
        print(f"🔗 Link: https://youtu.be/{video_id}")

        if os.path.exists(thumbnail_path):
            choice_thumb = input("🖼️ Kya aap custom thumbnail upload karna chahte hain? (y/n): ").strip().lower()
            if choice_thumb == 'y':
                try:
                    youtube.thumbnails().set(videoId=video_id, media_body=MediaFileUpload(thumbnail_path)).execute()
                    print("✅ Thumbnail uploaded successfully!")
                except Exception as thumb_err:
                    print(f"\n⚠ Thumbnail Error: YouTube channel phone verified nahi hai. Verify here: https://www.youtube.com/verify\n")
            else:
                print("⏭️ Thumbnail upload skipped.")

    except Exception as e:
        print(f"❌ YouTube Upload Error: {e}")

def main():
    start_time = time.time()
    initialize_gemini()

    download_source()
    breakdowns, full_transcript_text = process_adaptive_transcript()

    loop = asyncio.new_event_loop()
    asyncio.set_event_loop(loop)
    loop.run_until_complete(generate_voiceover_audio(breakdowns))
    loop.close()

    render_segmented_video(breakdowns)
    video_title = generate_metadata(full_transcript_text)

    upload_to_youtube(
        video_path=OUTPUT_VIDEO,
        thumbnail_path=THUMBNAIL,
        title=video_title,
        transcript_text=full_transcript_text,
        privacy_status="private"
    )

    print("\n" + "="*70)
    print(f"🎉 PODCAST PIPELINE v14.20 COMPLETE! Total execution: {int(time.time() - start_time)} seconds")
    print("="*70)

if __name__ == "__main__":
    main()
```
